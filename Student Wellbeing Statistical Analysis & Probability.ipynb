```python
📊 Student Wellbeing Survey – Statistical Analysis Notebook

# Step 1: Import Libraries
import pandas as pd
import numpy as np
from scipy import stats

# Step 2: Load Dataset
df = pd.read_csv("student_wellbeing.csv")

# Step 3: Inspect Dataset
print(df.shape)
print(df.info())
df.head()
```

```python
1. Central Tendency (Mean, Median, Mode)

variables = ["Weekly_Study_Hours","Average_Sleep_Hours",
             "Daily_Screen_Time_Hours","Stress_Score",
             "Academic_Readiness_Score"]

for var in variables:
    mean = df[var].mean()
    median = df[var].median()
    mode = df[var].mode()[0]
    print(f"{var}: Mean={mean:.2f}, Median={median:.2f}, Mode={mode}")

Interpretation: These values summarize the “typical” student behavior for study, sleep, screen time, stress, and readiness.
```

```python
2. Dispersion (Range, Variance, Std Dev, Quartiles, IQR)

for var in variables:
    rng = df[var].max() - df[var].min()
    varnc = df[var].var()
    std = df[var].std()
    q1 = df[var].quantile(0.25)
    q3 = df[var].quantile(0.75)
    iqr = q3 - q1
    print(f"{var}: Range={rng:.2f}, Var={varnc:.2f}, Std={std:.2f}, Q1={q1:.2f}, Q3={q3:.2f}, IQR={iqr:.2f}")

Interpretation: Identify which variable has the largest standard deviation/IQR → greatest variability.
```

```python
3. Outlier Detection (IQR Method)

outlier_vars = ["Weekly_Study_Hours","Daily_Screen_Time_Hours",
                "Commute_Time_Minutes","Monthly_Discretionary_Spending"]

for var in outlier_vars:
    q1 = df[var].quantile(0.25)
    q3 = df[var].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5*iqr
    upper = q3 + 1.5*iqr
    outliers = df[(df[var] < lower) | (df[var] > upper)]
    print(f"{var}: Outliers={outliers.shape[0]}")
```

```python
4. Compare Mean & Median Before/After Outlier Removal

var = "Weekly_Study_Hours"
before_mean, before_median = df[var].mean(), df[var].median()

# Remove outliers
q1, q3 = df[var].quantile(0.25), df[var].quantile(0.75)
iqr = q3 - q1
lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
df_no_outliers = df[(df[var] >= lower) & (df[var] <= upper)]

after_mean, after_median = df_no_outliers[var].mean(), df_no_outliers[var].median()

print(f"Before: Mean={before_mean:.2f}, Median={before_median:.2f}")
print(f"After: Mean={after_mean:.2f}, Median={after_median:.2f}")

Interpretation: Outliers skew the mean more than the median.
```

```python
5. Probability Events
Define events:

A: Part_Time_Job = "Yes"

B: Stress_Score ≥ 7

C: Scholarship = "Yes"

D: Exercise ≥ 3 days/week

#code
n = len(df)

P_A = (df["Part_Time_Job"]=="Yes").mean()
P_B = (df["Stress_Score"]>=7).mean()
P_C = (df["Scholarship"]=="Yes").mean()
P_D = (df["Exercise_Days_Per_Week"]>=3).mean()

P_A_or_B = ((df["Part_Time_Job"]=="Yes") | (df["Stress_Score"]>=7)).mean()
P_A_and_B = ((df["Part_Time_Job"]=="Yes") & (df["Stress_Score"]>=7)).mean()
P_A_given_B = P_A_and_B / P_B
P_B_given_A = P_A_and_B / P_A

print(P_A, P_B, P_C, P_D, P_A_or_B, P_A_and_B, P_A_given_B, P_B_given_A)

Mutual Exclusivity: Year_of_Study=1 and Year_of_Study=4 cannot occur together → mutually exclusive.

Independence Check: Compare P(A and B) vs P(A)*P(B).
```

```python
6. Bayes’ Theorem Verification

P_B_given_A = P_A_and_B / P_A
P_B_given_notA = ((df["Part_Time_Job"]!="Yes") & (df["Stress_Score"]>=7)).mean() / (df["Part_Time_Job"]!="Yes").mean()

bayes_P_A_given_B = (P_B_given_A * P_A) / (P_B_given_A * P_A + P_B_given_notA * (1-P_A))
print("Bayes P(A|B):", bayes_P_A_given_B, "Direct:", P_A_given_B)
```

```python
7. Normal Distribution (Academic_Readiness_Score)

mu = df["Academic_Readiness_Score"].mean()
sigma = df["Academic_Readiness_Score"].std()

z_high = (df["Academic_Readiness_Score"].max() - mu)/sigma
z_low = (df["Academic_Readiness_Score"].min() - mu)/sigma

print("Mean:", mu, "Std Dev:", sigma)
print("Highest Z:", z_high, "Lowest Z:", z_low)

#Empirical Rule:

#~68% within 1σ

#~95% within 2σ

#~99.7% within 3σ

within_1 = ((df["Academic_Readiness_Score"] >= mu-sigma) & (df["Academic_Readiness_Score"] <= mu+sigma)).mean()
within_2 = ((df["Academic_Readiness_Score"] >= mu-2*sigma) & (df["Academic_Readiness_Score"] <= mu+2*sigma)).mean()
within_3 = ((df["Academic_Readiness_Score"] >= mu-3*sigma) & (df["Academic_Readiness_Score"] <= mu+3*sigma)).mean()

print(within_1, within_2, within_3)
```

```python
📝 Final Observations (Example 5)
Stress scores show higher variability than study hours, indicating diverse coping levels among students.

Outliers in screen time heavily skew averages, highlighting extreme device usage in a small group.

Part-time jobs and high stress are not independent — students with jobs are more likely to report stress ≥7.

Bayes’ theorem verification confirms consistency between direct and conditional probability calculations.

Academic readiness scores follow a near-normal distribution, with most students clustered within 2 standard deviations of the mean.
```
